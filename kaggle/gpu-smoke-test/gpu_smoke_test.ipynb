# Stickman Motion Studio — Kaggle GPU smoke test

This first run only verifies that Kaggle starts a GPU notebook and that PyTorch can see CUDA. It does not install ComfyUI or generate images yet.

```python
import os, platform, json, subprocess
print('Python:', platform.python_version())
print('Working directory:', os.getcwd())
subprocess.run(['nvidia-smi'], check=False)
```

```python
import torch
report = {
    'torch_version': torch.__version__,
    'cuda_available': torch.cuda.is_available(),
    'cuda_version': torch.version.cuda,
    'gpu_count': torch.cuda.device_count(),
    'gpus': [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())],
}
print(json.dumps(report, indent=2))
assert torch.cuda.is_available(), 'Kaggle GPU is not visible to PyTorch. Check notebook accelerator settings.'
x = torch.randn((1024, 1024), device='cuda')
y = x @ x
print('GPU matrix test:', tuple(y.shape), 'on', y.device)
```

```python
from pathlib import Path
Path('/kaggle/working/gpu-smoke-test.json').write_text(json.dumps(report, indent=2))
print('Wrote /kaggle/working/gpu-smoke-test.json')
```
